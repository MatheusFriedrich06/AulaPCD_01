```python
nomeAluno = str (input ("informe o seu nome: "))

A = float (input ("Informe a terceira nota: "))
B = float (input ("Informe a segunda nota: "))
C = float (input ("Informe a terceira nota: "))

valorMedia = (A + B + C) / 3

print(f"Aluno: {nomeAluno}")
print(f"As notas são: {A}, {B}, {C}")
print(f"A média das notas ficaria: {valorMedia}")
```
